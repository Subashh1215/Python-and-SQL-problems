```python
n = int(input("Enter number of integers: "))
numbers = list(map(int, input("Enter the numbers: ").split()))

positive = negative = zero = 0
even = odd = 0
both = only3 = only5 = neither = 0

for num in numbers[:n]:

    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    else:
        zero += 1

    if num != 0:
        if num % 2 == 0:
            even += 1
        else:
            odd += 1

    if num % 3 == 0 and num % 5 == 0:
        both += 1
    elif num % 3 == 0:
        only3 += 1
    elif num % 5 == 0:
        only5 += 1
    else:
        neither += 1

print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
print("Even:", even)
print("Odd:", odd)
print("Divisible by both 3 and 5:", both)
print("Divisible only by 3:", only3)
print("Divisible only by 5:", only5)
print("Divisible by neither:", neither)
```

```python
invalid_attempts = 0

while True:
    try:
        value = int(input("Enter an integer between 1 and 100: "))
    except ValueError:
        print("Invalid input")
        invalid_attempts += 1
        continue

    if value < 1 or value > 100:
        print("Invalid input")
        invalid_attempts += 1
        continue

    break

print("Accepted value:", value)
print("Invalid attempts:", invalid_attempts)
```

```python
while True:
    print("\n--- Unit Converter ---")
    print("1. Celsius to Fahrenheit")
    print("2. Kilometres to Miles")
    print("3. Kilograms to Pounds")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        celsius = float(input("Enter Celsius: "))
        fahrenheit = (celsius * 9 / 5) + 32
        print(f"Fahrenheit: {fahrenheit:.2f}")

    elif choice == "2":
        km = float(input("Enter kilometres: "))
        miles = km * 0.621371
        print(f"Kilometres to Miles: {miles:.2f}")

    elif choice == "3":
        kg = float(input("Enter kilograms: "))
        pounds = kg * 2.20462
        print(f"Kilograms to Pounds: {pounds:.2f}")

    elif choice == "4":
        print("Exit")
        break

    else:
        print("Invalid choice")
        continue
```

```python
n = int(input("Enter a number: "))
skipped = 0

for i in range(1, 21):
    result = n * i

    if result > 100:
        break

    if result % 3 == 0:
        skipped += 1
        continue

    print(f"{n} x {i} = {result}")

print("Skipped results:", skipped)
```

```python
password = input("Enter password: ")

uppercase = False
lowercase = False
digit = False
special = False

for char in password:
    if char.isupper():
        uppercase = True
    elif char.islower():
        lowercase = True
    elif char.isdigit():
        digit = True
    else:
        special = True

length_valid = len(password) >= 8

if uppercase and lowercase and digit and special and length_valid:
    print("Valid password")
else:
    print("Invalid password")

print("Uppercase:", "Present" if uppercase else "Missing")
print("Lowercase:", "Present" if lowercase else "Missing")
print("Digit:", "Present" if digit else "Missing")
print("Special character:", "Present" if special else "Missing")
print("Minimum length:", "Satisfied" if length_valid else "Not satisfied")
```

```python
n = int(input("Enter the size: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
L, R = map(int, input("Enter L and R: ").split())

found = False

for num in range(max(0, L), R + 1):
    i = 0

    while i * i < num:
        i += 1

    if i * i == num:
        print("First perfect square:", num)
        found = True
        break

if not found:
    print("No perfect square found in the range.")
```

```python
secret = 8
attempts = 0
guessed = False

while attempts < 5:
    guess = int(input("Guess the number: "))
    attempts += 1

    if guess == secret:
        print("Correct")
        guessed = True
        break
    elif guess < secret:
        print("Too Low")
    else:
        print("Too High")

if not guessed:
    print("Game Over")
    
```

```python
n = int(input("Enter number of elements: "))
numbers = list(map(int, input("Enter the numbers: ").split()))

if n == 0:
    print("Sequence is empty")
else:
    current_value = numbers[0]
    current_length = 1

    best_value = numbers[0]
    best_length = 1

    for i in range(1, min(n, len(numbers))):
        if numbers[i] == current_value:
            current_length += 1
        else:
            current_value = numbers[i]
            current_length = 1

        if current_length > best_length:
            best_length = current_length
            best_value = current_value

    print("Value =", best_value)
    print("Length =", best_length)
```

```python
normal = 0
large = 0
suspicious = 0
total = 0

while True:
    amount = int(input("Enter transaction amount (-1 to stop): "))

    if amount == -1:
        break

    if amount == 0:
        pass

    elif amount < 0:
        continue

    elif amount <= 1000:
        normal += 1
        total += amount

    elif amount <= 5000:
        large += 1
        total += amount

    else:
        suspicious += 1
        total += amount

        if suspicious == 3:
            break

print("Normal transactions:", normal)
print("Large transactions:", large)
print("Suspicious transactions:", suspicious)
print("Total valid amount:", total)
```
