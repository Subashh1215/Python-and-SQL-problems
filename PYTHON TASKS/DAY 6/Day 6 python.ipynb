```python
stock = {'LED': 20, 'RESISTOR': 50, 'SENSOR': 4}
requested = ['LED', 'SENSOR', 'SWITCH']

for item in requested:
    if item in stock:
        print(f"{item}: {stock[item]} available")
    else:
        print(f"{item}: unavailable")
```

```python
A = {101, 102, 103, 105}
B = {102, 104, 105, 106}

both_courses = sorted(A & B)
only_A = sorted(A - B)
either_course = sorted(A | B)

print("Both courses:", both_courses)
print("Only Course A:", only_A)
print("Either course:", either_course)
```

```python
names = ['Anu', 'Arun', 'Bala', 'Banu', 'Cathy', 'Asha']

groups = {}

for name in names:
    letter = name[0].upper()

    if letter not in groups:
        groups[letter] = []

    groups[letter].append(name)

print(groups)
```

```python
store_a = {'LED': 20, 'Cable': 10, 'Board': 3}
store_b = {'Cable': 5, 'Sensor': 7, 'LED': 8}

combined = store_a.copy()

for item, quantity in store_b.items():
    combined[item] = combined.get(item, 0) + quantity

print("Combined inventory:", combined)
```

```python
sales = {
    'Sensor': [4, 6, 5],
    'Cable': [10, 2, 3],
    'Board': [8, 7],
    'LED': [5, 5, 5]
}

totals = {}

for product, quantities in sales.items():
    totals[product] = sum(quantities)

ranked = sorted(totals.items(), key=lambda item: (-item[1], item[0]))

for product, total in ranked:
    print(f"{product}: {total}")
```

```python
permissions = {
    'admin': {'read', 'write', 'delete'},
    'operator': {'read', 'write'},
    'auditor': {'read', 'export'}
}

user_permissions = list(permissions.values())

common = set.intersection(*user_permissions)
all_permissions = set.union(*user_permissions)

print("Common permissions:", sorted(common))
print("All permissions:", sorted(all_permissions))
```

```python
nums = [1, 1, 1]
k = 2

prefix_sum = 0
count = 0
prefix_counts = {0: 1}

for num in nums:
    prefix_sum += num

    if prefix_sum - k in prefix_counts:
        count += prefix_counts[prefix_sum - k]

    prefix_counts[prefix_sum] = prefix_counts.get(prefix_sum, 0) + 1

print("Count:", count)
```

```python
from collections import Counter

s = 'ADOBECODEBANC'
t = 'ABC'

if not t or not s:
    print("Minimum window: ''")
else:
    required = Counter(t)
    window = {}

    left = 0
    formed = 0
    required_chars = len(required)

    min_length = float('inf')
    min_start = 0

    for right, char in enumerate(s):
        window[char] = window.get(char, 0) + 1

        if char in required and window[char] == required[char]:
            formed += 1

        while formed == required_chars:
            current_length = right - left + 1

            if current_length < min_length:
                min_length = current_length
                min_start = left

            left_char = s[left]
            window[left_char] -= 1

            if (left_char in required and
                    window[left_char] < required[left_char]):
                formed -= 1

            left += 1

    if min_length == float('inf'):
        result = ''
    else:
        result = s[min_start:min_start + min_length]

    print("Minimum window:", repr(result))
```
